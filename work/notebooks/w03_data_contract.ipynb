## 2. Fields: feature / label / context / excluded

### Features
- `search_volume` — keyword demand known at the decision moment.
- `competition` — keyword competition known at the decision moment.
- `backlinks` — existing backlink count known at the decision moment.
- `word_count` — current content length known at the decision moment.
- `content_age_days` — calculated from content creation date and decision date.

### Label
- Future organic sessions — the future outcome proxy used to rank content.

### Context
- `client_hash_id` — identifies the client.
- `content_hash_id` — identifies the content item.
- `report_date` — identifies the observation date.
- `gsc_data_available` — indicates whether GSC data is available.
- `ga4_data_available` — indicates whether GA4 data is available.

### Excluded
- Future-period performance fields — excluded because they would reveal information from the outcome window and cause leakage.

## 4. Data limits

This slice cannot tell us why a content item performed well or poorly. The history is also unbalanced because some clients may have GSC or GA4 data available for only part of the period. Therefore, the results are directional and should be used for decision-support rather than treated as causal conclusions. Future outcome windows can also overlap with the period used to create features, so the time boundary must be kept explicit.

### Five features and availability

| Feature | Available when? |
|---|---|
| `search_volume` | Knowable at the decision moment because it comes from the keyword data available before the future outcome window. |
| `competition` | Knowable at the decision moment because keyword competition is an existing search-market attribute. |
| `backlinks` | Knowable at the decision moment because the existing backlink count can be observed before the outcome period. |
| `word_count` | Knowable at the decision moment because the content's current word count is known before future performance is observed. |
| `content_age_days` | Knowable at the decision moment because it is calculated from the content creation date and the decision date. |

```python
march_path = "hf://datasets/FlyRank/internship-warehouse/fact_content_daily_performance/month=2026-03/data_0.parquet"

feature_frame = conn.sql(f"""
SELECT
    f.client_hash_id,
    f.content_hash_id,
    f.report_date,

    c.search_volume,
    c.competition,
    c.backlinks,
    c.word_count,

    DATE_DIFF(
        'day',
        c.content_created_date,
        f.report_date
    ) AS content_age_days

FROM read_parquet('{march_path}') AS f

LEFT JOIN read_parquet(
    'hf://datasets/FlyRank/internship-warehouse/dim_content.parquet'
) AS c
    ON f.client_hash_id = c.client_hash_id
   AND f.content_hash_id = c.content_hash_id

WHERE f.gsc_data_available IS TRUE
   OR f.ga4_data_available IS TRUE
""").df()

print("Feature frame shape:", feature_frame.shape)
feature_frame.head()
```

```python
april_path = "hf://datasets/FlyRank/internship-warehouse/fact_content_daily_performance/month=2026-04/data_0.parquet"

label_frame = conn.sql(f"""
SELECT
    client_hash_id,
    content_hash_id,
    SUM(sessions_organic) AS future_organic_sessions
FROM read_parquet('{april_path}')
GROUP BY
    client_hash_id,
    content_hash_id
""").df()

print("Label frame shape:", label_frame.shape)
label_frame.head()
```

```python
model_frame = feature_frame.merge(
    label_frame,
    on=["client_hash_id", "content_hash_id"],
    how="left"
)

model_frame["future_organic_sessions"] = (
    model_frame["future_organic_sessions"].fillna(0)
)

print("Model frame shape:", model_frame.shape)
model_frame.head()
```

```python
import numpy as np
import pandas as pd

feature_cols = [
    "search_volume",
    "competition",
    "backlinks",
    "word_count",
    "content_age_days"
]

score_df = model_frame[
    feature_cols + ["future_organic_sessions"]
].dropna().copy()

# Limit computation to a reproducible sample
score_sample = score_df.sample(
    n=min(100000, len(score_df)),
    random_state=42
)

correlations = (
    score_sample.corr(numeric_only=True)["future_organic_sessions"]
    .drop("future_organic_sessions")
    .sort_values(key=abs, ascending=False)
)

honest_score = correlations.abs().max()

print("Honest baseline score:", round(honest_score, 4))
print("\nFeature correlations:")
print(correlations)
```

```python
leak_df = score_sample.copy()

# Deliberate leakage: copy the future label into a feature
leak_df["leaked_future_organic_sessions"] = (
    leak_df["future_organic_sessions"]
)

leak_score = leak_df[
    ["leaked_future_organic_sessions", "future_organic_sessions"]
].corr().iloc[0, 1]

print("Leaked score:", round(leak_score, 4))
```

```python
# Remove the deliberately leaked label-derived column
clean_df = leak_df.drop(
    columns=["leaked_future_organic_sessions"]
)

clean_correlations = (
    clean_df.corr(numeric_only=True)["future_organic_sessions"]
    .drop("future_organic_sessions")
    .sort_values(key=abs, ascending=False)
)

honest_score_final = clean_correlations.abs().max()

print("Honest score after removing leakage:",
      round(honest_score_final, 4))

print("\nLeakage lesson:")
print("The leaked score was artificially perfect because the label itself was used as a feature.")
print("The final honest score excludes the label-derived column.")
```
